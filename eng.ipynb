```python
import math
class Value: 
    def __init__(self, data, children=(), op=''):
        self.data = data
        self.prev = set(children) # Children in the previous layer
        self.prev_op = op # Operation that the children Values used to create the current Value object. 
        self.grad = 0
        self._backward = lambda: None

    def __add__(self, other):  # assumes 'other' is also an object of the Value class. 
        out = Value(self.data + self.other, (self, other), '+')

        def backward(): 
            self.grad = 1 * out.grad # The ' * out.grad is so that the chain rule kicks in. Draw it out a NN to see it.'
            other.grad = 1 * out.grad
        out._backward = backward # Stores the calculations of the function (It stores the blueprint of the function). No parenthesis for backward because that would run the function immediately. 
        return out 


    def __mult__(self, other): # Adding in underscores allows python to understand that you want to use the "*" symbol instead of typing a.__mult__(b)
        out = Value(self.data * self.other, (self, other), '*')
        def backward(): 
            self.grad = other.data * out.grad 
            other.grad = self.data * out.grad
        out._backward = backward # Stores the calculations of the function. No parenthesis for backward because that would run the function immediately. 
        return out

    def tanh(self): 
        tanh_val = (math.e**self.data - math.e**(-self.data)) / (math.e**self.data + math.e**(-self.data))
        out = Value(tanh_val, (self,), 'tanh')
        def backward(): 
            self.grad = 4 / (math.exp(self.data) + math.exp(-self.data)) ** 2

        out._backward = backward # Stores the calculations of the function. No parenthesis for backward because that would run the function immediately. 
        return out

```
